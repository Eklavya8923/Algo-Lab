# Shortest Path Using Dijkstra Algorithm in Python
Problem Description:
Given a weighted graph with non negative edge weights. Find the shortest distance from a source vertex to all other vertices using Dijkstra algorithm

#Theory
Dijkstra algorithm finds shortest distance from one source node to all other nodes in a weighted graph with non negative weights. Dictionary structure stores nodes as keys and neighbor nodes with weights as values.

##Algorithm Steps

1. Represent graph using dictionary.
2. Set distance of every node as infinity.
3. Set distance of source node as 0.
4. Select node with smallest distance among unvisited nodes.
5. Update distance of neighbor nodes.
6. Repeat until all nodes get visited.

```python
def dijkstra(graph, start):
    distances = {}
    visited = []
    previous_nodes = {}

    for node in graph:
        distances[node] = float('inf')
        previous_nodes[node] = None

    distances[start] = 0

    while len(visited) < len(graph):
        min_node = None
        min_distance = float('inf')

        for node in graph:
            if node not in visited and distances[node] < min_distance:
                min_distance = distances[node]
                min_node = node

        if min_node is None:
            break

        visited.append(min_node)

        for neighbor, weight in graph[min_node].items():
            new_distance = distances[min_node] + weight

            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                previous_nodes[neighbor] = min_node

    return distances, previous_nodes

def get_path_string(previous_nodes, start, target):
    path = []
    current_node = target

    while current_node is not None:
        path.insert(0, current_node)
        current_node = previous_nodes[current_node]

    if path and path[0] == start:
        return " -> ".join(path)
    return "No path"

def get_user_graph():
    user_graph = {}
    num_nodes = int(input("Total nodes: "))

    for _ in range(num_nodes):
        node = input("Node name: ").strip()
        user_graph[node] = {}
        num_neighbors = int(input(f"Neighbors for {node}: "))

        for _ in range(num_neighbors):
            neighbor = input("Neighbor name: ").strip()
            weight = int(input("Distance: "))
            user_graph[node][neighbor] = weight

    return user_graph

print("--- Task 1: User Input Graph ---")
# user_graph = get_user_graph()
# print(user_graph)

print("\n--- Task 3: Graph with 6 Nodes ---")
my_graph = {
    'A': {'B': 4, 'C': 2},
    'B': {'A': 4, 'C': 5, 'D': 10, 'E': 3},
    'C': {'A': 2, 'B': 5, 'D': 3, 'F': 8},
    'D': {'B': 10, 'C': 3, 'E': 1},
    'E': {'B': 3, 'D': 1, 'F': 4},
    'F': {'C': 8, 'E': 4}
}
print("Graph loaded.")

print("\n--- Task 2 & 4: Shortest Paths from Source 'A' ---")
start_node = 'A'
distances, previous_nodes = dijkstra(my_graph, start_node)

for target_node in my_graph:
    route = get_path_string(previous_nodes, start_node, target_node)
    print(f"Path to {target_node}: {route} | Distance = {distances[target_node]}")

print("\n--- Task 2 & 4: Shortest Paths from Source 'E' ---")
start_node = 'E'
distances, previous_nodes = dijkstra(my_graph, start_node)

for target_node in my_graph:
    route = get_path_string(previous_nodes, start_node, target_node)
    print(f"Path to {target_node}: {route} | Distance = {distances[target_node]}")
```
