# MATRIX CHAIN MULTIPLICATION Dynamic Programming with table filling.
Problem Description:
Find minimum number of scalar multiplications needed to multiply a chain of matrices..

#Theory
Cost of multiplying two matrices:
If A is p×q and B is q×r, cost = p × q × r

Dynamic Programming idea:
Break problem into smaller subproblems.
Store results in a table to avoid recomputation.

m[i][j] = minimum cost to multiply Ai to Aj

## Algorithm
1. Take array p[] of size n
2. Create table m[n][n]
3. Set m[i][i] = 0
4. Fill table diagonally
5. For each chain length L:
    Compute m[i][j]
    Try all possible k between i and j
6. Choose minimum cost
7. Return m[1][n-1]

```python
import sys

def matrix_chain_multiplication(p):
    n = len(p)


    m = [[0 for _ in range(n)] for _ in range(n)]
    max_m = [[0 for _ in range(n)] for _ in range(n)]
    s = [[0 for _ in range(n)] for _ in range(n)]


    for L in range(2, n):
        for i in range(1, n - L + 1):
            j = i + L - 1
            m[i][j] = sys.maxsize
            max_m[i][j] = 0


            for k in range(i, j):

                multiplications = p[i-1] * p[k] * p[j]


                cost_min = m[i][k] + m[k+1][j] + multiplications
                if cost_min < m[i][j]:
                    m[i][j] = cost_min
                    s[i][j] = k


                cost_max = max_m[i][k] + max_m[k+1][j] + multiplications
                if cost_max > max_m[i][j]:
                    max_m[i][j] = cost_max

    return m, s, max_m


def get_optimal_parens(s, i, j):
    if i == j:
        return f"A{i}"
    else:
        left = get_optimal_parens(s, i, s[i][j])
        right = get_optimal_parens(s, s[i][j] + 1, j)
        return f"({left} {right})"


p = [5, 10, 3, 12, 5, 50, 6]
n = len(p)


m, s, max_m = matrix_chain_multiplication(p)

min_cost = m[1][n-1]
worst_cost = max_m[1][n-1]
saved = worst_cost - min_cost
optimal_structure = get_optimal_parens(s, 1, n-1)

print("=== LAB EXERCISE RESULTS ===")
print(f"1. Input dimensions array: p = {p}")
print(f"2. Minimum scalar multiplications: {min_cost}")
print(f"3. Optimal Parenthesization: {optimal_structure}")
print("4. Multiplications Saved:")
print(f"   -> Worst-case groupings cost: {worst_cost}")
print(f"   -> Optimal groupings cost:    {min_cost}")
print(f"   -> Total operations saved:    {saved}")
```
